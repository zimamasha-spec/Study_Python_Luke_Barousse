```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.index
```

```python
df.index.name = 'job_index'
```

```python
df.head()
```

```python
df.index.dtype
```

# reset_index()

```python
df_ukr = df[df['job_country']=="Ukraine"]

df_ukr.reset_index(inplace = True)

df_ukr
```

# set_index()

```python
df_ukr.set_index('job_index', inplace = True)
```

```python
df_ukr
```

# sort_index()

```python
median_pivot = df_ukr.pivot_table(values = 'salary_year_avg', index = 'job_title_short', aggfunc = ['median', 'min', 'max'])
```

```python
median_pivot
```

```python
median_pivot.sort_values(by = ('median', 'salary_year_avg'), ascending=False)
```

```python
median_pivot.sort_index(inplace=True)
median_pivot
```
